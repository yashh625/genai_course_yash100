```python
from google.colab import userdata
from openai import OpenAI

# 1. Fetch your OpenRouter key safely from Colab Secrets
OPENROUTER_API_KEY = userdata.get('langchain_genaicourse')

# 2. Initialize the OpenAI client properly
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=OPENROUTER_API_KEY,
)

# 3. Setup your prompt
prompt_text = "Tell me about yourself and what you can do"

# 4. Use a completely new variable name to bypass Colab's cached memory string
completion_result = client.chat.completions.create(
    model="openrouter/free",
    messages=[
        {"role": "user", "content": prompt_text}
    ]
)

# 5. Extract and print the final generated message
print(completion_result.choices[0].message.content)
```
